```python
# importing required libraries
import pandas as pd
```

```python
t = pd.read_csv("../data/raw/traffic_sensor_log.csv")
w = pd.read_csv("../data/raw/weather_observations.csv")
c = pd.read_csv("../data/raw/calendar_events.csv")
```

```python
# shape
print("traffic rows:", len(t))
print("unique road_id:", t["road_id"].nunique())
```

```python
# duplicates — key-level, not just exact-row
dupe_count = t.duplicated(subset=["road_id", "date", "time"]).sum()
print("key-level duplicates:", dupe_count)
```

```python
# missing values per column
print(t.isnull().sum())
```

```python
# impossible values — rule-based outliers
print("negative traffic_volume:", (t["traffic_volume"] < 0).sum())
print("avg_speed > 200:", (t["avg_speed"] > 200).sum())
print("occupancy > 100:", (t["occupancy"] > 100).sum())
```

```python
# blank target
print("blank congestion_level %:", t["congestion_level"].isnull().mean() * 100)

```
